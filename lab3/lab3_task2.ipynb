## Import libraries and load image


```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

image = cv2.imread("images/cameraman.jpg",0)
```

## 1.Negative transformation


```python
negative =255 - image

plt.imshow(negative,cmap="gray")
plt.axis("off")
plt.title("negative image")
```

# 2 Intensity Tranformation


```python
# Convert to float
img_float = np.float32(image)

# Choose appropriate constant
c = 255 / np.log(1 + np.max(img_float))

# Apply log transformation
log_img = c * np.log(1 + img_float)

# Convert back to uint8
log_img = np.uint8(log_img)

cv2.imshow("Log Transformed Image", log_img) 
cv2.waitKey(0)
```

# 3 Power Law Transformation

```python
# Normalize image to range [0,1]
img_norm = image / 255.0

gamma = 0.5   # increases contrast
c = 1

power_img = c * np.power(img_norm, gamma)

# Scale back to [0,255]
power_img = np.uint8(power_img * 255)

cv2.imshow("Power Law (Gamma=0.5)", power_img)
cv2.waitKey(0)

cv2.destroyAllWindows()
```
