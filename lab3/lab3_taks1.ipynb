# 1.Increase the size of image

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

image = cv2.imread("images/cameraman.jpg")

# the scale set to 1.5
scale_x = 1.5
scale_y = 1.5
new_image = cv2.resize(image,None,fx=scale_x,fy=scale_y,interpolation=cv2.INTER_LINEAR)

plt.imshow(cv2.cvtColor(new_image, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.show()
```

# 2.Rotate the image by 120 degree


```python
(h, w) = image.shape[:2]
center = (w // 2, h // 2)

rotation_matrix = cv2.getRotationMatrix2D(center, 120, 1.0)
rotated = cv2.warpAffine(image, rotation_matrix, (w, h))

plt.imshow(cv2.cvtColor(rotated, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.show()
```

# 3.Shear transformation


```python
shear_factor = 0.5

shear_matrix = np.array([
    [1, shear_factor, 0],
    [0, 1, 0]
], dtype=np.float32)

sheared = cv2.warpAffine(image, shear_matrix, (int(w + shear_factor * h), h))

plt.imshow(cv2.cvtColor(sheared, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.show()
```
