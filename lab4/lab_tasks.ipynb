### Task 1

#### Original image

```python
import matplotlib.pyplot as plt
import numpy as np
from skimage import data, exposure

# Load moon image
img = data.moon()

# Display original image
plt.imshow(img, cmap='gray')
plt.axis('off')
plt.title('Original Image')
plt.show()
```

### Rescaled Image

```python

# Load moon image
img = data.moon()

# Compute 3rd and 80th percentiles
p3, p80 = np.percentile(img, (3, 80))

# Contrast stretching
img_rescale = exposure.rescale_intensity(img, in_range=(p3, p80))

# Display image and histogram
fig = plt.figure(figsize=(10,7))

fig.add_subplot(2,1,1)
plt.imshow(img_rescale, cmap='gray')
plt.axis('off')
plt.title('Rescaled Image')

fig.add_subplot(2,1,2)
plt.hist(img_rescale.flat, bins=256, range=(0,255))
plt.xlabel('Gray level values')
plt.ylabel('Number of pixels')
plt.title('Histogram after Rescaling')

plt.show()
```

### task 2

```python
# Histogram Equalization
img_equalized = exposure.equalize_hist(img)

# Display image and histogram
fig = plt.figure(figsize=(10,7))

fig.add_subplot(2,1,1)
plt.imshow(img_equalized, cmap='gray')
plt.axis('off')
plt.title('Histogram Equalized Image')

fig.add_subplot(2,1,2)
plt.hist(img_equalized.flat, bins=256, range=(0,1))
plt.xlabel('Gray level values')
plt.ylabel('Number of pixels')
plt.title('Histogram after Equalization')

plt.show()
```

### task 3

```python
from skimage.exposure import match_histograms

# Load images
source = data.chelsea()   # image to be transformed
reference = data.rocket() # reference image

# Apply histogram matching
matched = match_histograms(source, reference, channel_axis=-1)

# Display results
fig, axes = plt.subplots(1, 3, figsize=(15,5))

axes[0].imshow(source)
axes[0].set_title('Source (Chelsea)')
axes[0].axis('off')

axes[1].imshow(reference)
axes[1].set_title('Reference (Rocket)')
axes[1].axis('off')

axes[2].imshow(matched)
axes[2].set_title('Histogram Matched')
axes[2].axis('off')

plt.show()
```
