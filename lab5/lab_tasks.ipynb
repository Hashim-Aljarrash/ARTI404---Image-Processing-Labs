## Task 1 - Convolve with 7x7 box filter

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read image
image = cv2.imread('images/Parrot.png')
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# Create 7x7 box filter kernel
kernel = np.ones((7, 7), np.float32) / 49

# Apply convolution
smoothed = cv2.filter2D(image, -1, kernel)

# Plot results
fig, ax = plt.subplots(1, 2, figsize=(10, 5))

ax[0].imshow(image)
ax[0].set_title("Original Image")
ax[0].axis("off")

ax[1].imshow(smoothed)
ax[1].set_title("7x7 Box Filter (Smoothed)")
ax[1].axis("off")

plt.show()
```

## Task 2 - Gaussian Filter (5x5) and (21x21)

```python

# Read image (you can change the file name if desired)
image = cv2.imread('images/Parrot.png')
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# Apply Gaussian filters
gaussian_5 = cv2.GaussianBlur(image, (5, 5), 0)
gaussian_21 = cv2.GaussianBlur(image, (21, 21), 0)

# Plot results
fig, ax = plt.subplots(1, 3, figsize=(15, 5))

ax[0].imshow(image)
ax[0].set_title("Original Image")
ax[0].axis("off")

ax[1].imshow(gaussian_5)
ax[1].set_title("Gaussian 5x5")
ax[1].axis("off")

ax[2].imshow(gaussian_21)
ax[2].set_title("Gaussian 21x21")
ax[2].axis("off")

plt.show()
```

## Task 3 - Laplacian Sharpening

```python
# 1. Read the image
image = cv2.imread('images/Parrot.png')
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# 2. Convert to float32
image_float = image.astype(np.float32) / 255.0

# 3. Apply Laplacian (3x3)
laplacian = cv2.Laplacian(image_float, cv2.CV_32F, ksize=3)

# 4. Sharpen the image
sharpened = np.clip(image_float - laplacian, 0, 1)

# 5. Plot results
fig, ax = plt.subplots(1, 3, figsize=(15, 5))

ax[0].imshow(image)
ax[0].set_title("Original Image")
ax[0].axis("off")

ax[1].imshow(np.clip(laplacian + 0.5, 0, 1))  # shifted for visibility
ax[1].set_title("Laplacian Result")
ax[1].axis("off")

ax[2].imshow(sharpened)
ax[2].set_title("Sharpened Image")
ax[2].axis("off")

plt.show()
```
